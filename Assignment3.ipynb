```python

import heapq

def GBFS(graph, heuristic, start, goal):
    pq = []                  # Priority queue
    visited = set()          # Stores visited nodes
    parent = {}              # Stores parent of each node

    # Insert start node with its heuristic value
    heapq.heappush(pq, (heuristic[start], start))

    while pq:
        h, current = heapq.heappop(pq)  # Get node with lowest heuristic

        if current in visited:
            continue

        print("Visited:", current)
        visited.add(current)

        # Check if goal is reached
        if current == goal:
            path = []
            while current in parent:
                path.append(current)
                current = parent[current]
            path.append(start)
            path.reverse()   # Reverse to get start-to-goal path
            return path

        # Explore all neighbouring nodes
        for neighbour in graph[current]:
            if neighbour not in visited:
                if neighbour not in parent:
                    parent[neighbour] = current

                # Add neighbour with its heuristic value
                heapq.heappush(pq, (heuristic[neighbour], neighbour))

    return None              # No path found


graph = {}
heuristic = {}

# Input vertices and their heuristic values
vertices = int(input("Enter number of vertices: "))

for i in range(vertices):
    node = input("Enter node name: ").upper()
    graph[node] = []
    heuristic[node] = int(input("Enter heuristic value of " + node + ": "))

# Input edges between vertices
edges = int(input("Enter the number of edges: "))

print("Enter each edge (source destination):")

for i in range(edges):
    u = input("Enter the source node: ").upper()
    v = input("Enter the destination node: ").upper()
    graph[u].append(v)
    graph[v].append(u)  # For undirected graph

# Input start and goal nodes
start = input("Enter the start node: ").upper()
end = input("Enter the goal node: ").upper()

# Check whether start and goal nodes exist
if start not in graph or end not in graph:
    print("Invalid start or goal node")
else:
    path = GBFS(graph, heuristic, start, end)

    # Display the final path
    if path:
        print("Path Found:", " -> ".join(path))
    else:
        print("No Path Found")
```

```python
# GBFS is an informed search algorithm that selects the next node based on the lowest heuristic value to reach the goal.f(n)=h(n)
# A heuristic function estimates the distance or cost from the current node to the goal node.
# visited = set() : It stores already visited nodes and prevents revisiting them.
# heapq : To implement a priority queue that selects the node with the lowest heuristic value first.
# priority queue : A data structure in which elements are processed according to their priority rather than insertion order.
```
