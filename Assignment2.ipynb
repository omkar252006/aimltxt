```python

from collections import deque

# Breadth First Search (BFS)
def bfs(graph, src, goal):
    visited = set()
    queue = deque([src])  # FIFO queue
    visited.add(src)
    result = []

    while queue:
        curr = queue.popleft()  # Remove first node
        result.append(curr)
        print("BFS :", result)

        if curr == goal:
            print("Goal Reached")
            return

        # Visit all unvisited connected nodes
        for node in graph[curr]:
            if node not in visited:
                visited.add(node)
                queue.append(node)
                print("Queue :", list(queue))

    print("Goal Not Found")


# Depth First Search (DFS)
def dfs(graph, src, goal):
    visited = set()
    stack = [src]  # LIFO stack
    visited.add(src)
    result = []

    while stack:
        curr = stack.pop()  # Remove last node
        result.append(curr)
        print("DFS:", result)

        if curr == goal:
            print("Goal Found")
            return

        # Visit all unvisited connected nodes
        for node in graph[curr]:
            if node not in visited:
                visited.add(node)
                stack.append(node)
                print("Stack:", stack)

    print("Goal Not Found")


# Input number of nodes and create graph
cnt = int(input("Enter number of nodes: "))
graph = {}

for i in range(cnt):
    node = input(f"Enter node {i + 1}: ").strip()
    links = []
    total = int(input(f"Enter number of connected nodes for {node}: "))

    # Input connected nodes
    for j in range(total):
        nxt = input(f"Enter connected node {j + 1}: ").strip()
        links.append(nxt)

    graph[node] = links

print("\nGraph:", graph)

# Input source and destination nodes
src = input("\nEnter start node: ").strip()
goal = input("Enter goal node: ").strip()

# Validate source and goal nodes
if src not in graph:
    print("Start node does not exist.")
elif goal not in graph:
    print("Goal node does not exist.")
else:
    choice = 0

    # Display menu until user exits
    while choice != 3:
        print("\n-------------------------")
        print("1. Breadth First Search")
        print("-------------------------")
        print("2. Depth First Search")
        print("-----------------------")
        print("3. Exit")
        print("-------------------------")

        choice = int(input("Enter your choice: "))

        match choice:
            case 1:
                print("\nBreadth First Search Traversal")
                bfs(graph, src, goal)
            case 2:
                print("\nDepth First Search Traversal")
                dfs(graph, src, goal)
            case 3:
                print("Program Ended")
            case _:
                print("Invalid Choice")
```

```python
# A graph is a non-linear data structure consisting of vertices (nodes) and edges (connections) between them.
# Breadth First Search (BFS) is a graph traversal algorithm that explores nodes level by level, visiting nearby nodes first.
# Depth First Search (DFS) is a graph traversal algorithm that explores one path as deeply as possible before backtracking.
# BFS uses a Queue (FIFO).
# DFS uses a Stack (LIFO).
# Shortest Path: BFS (unweighted graph).
# Visited Set: Prevents repeated visits and infinite traversal through cycles.
```
