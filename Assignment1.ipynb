```python
while True : 
    print("\n----- Menu -----")
    print("1. Automatic Light")
    print("2. Air Conditioner")
    print("3. Door Security")
    print("4. Water Tank")
    print("5. TV Control")
    print("6. Exit System")

    choice = int(input("Enter Choice : "))

    match choice:
        case 1:
            people = input("Is person detected in the room (yes/no) : ").lower()
            darkness = input("Is the room dark (yes/no) : ").lower()

            # Control lights based on human presence and darkness
            if people == "no" and darkness == "yes" : 
                print("Lights remain OFF.")

            elif people == "no" and darkness == "no" :
                print("Lights are Turned OFF.")
                
                   
            elif darkness == "yes" and people == "yes" :
                print("Lights are Turned ON.")
            else:
                print("Lights remain ON.")

        case 2 : 
            temp = int(input("Enter Temperature : "))

            # Control AC based on room temperature
            if temp < 22:
                print("AC Turned OFF.")
            elif temp > 28 :
                print("AC Turned ON.")
            else :
                print("No change in AC Mode.")
        
        case 3 :
            security = input("Is any unauthorized person detected (yes/no) : ").lower()

            # Secure the door when an unauthorized person is detected
            if security == "yes" :
                print("Lock the door.")
            else :
                print("Unlock the door.")

        case 4:
            water_level = int(input("Enter Water Level : "))

            # Control water pump based on tank level
            if water_level < 0 or water_level > 100:
                print("Invalid Water Level entered.")
            elif water_level <20:
                print("Water pump turned ON.")
            elif water_level > 80:
                print("Water pump turned OFF.")
            else:
                print("Water pump in initial mode.")

        case 5:
            tv = input("Do you want to watch television (yes/no) : ").lower()

            # Control TV based on user preference
            if tv == "yes":
                print("TV Turned ON.")
            else:
                print("TV Turned OFF.")
                
        case 6:
            print("Exiting System.")
            break
            
        case _:
            print("Invalid choice entered.")
            
        
```
