```python
# Rule-Based Expert System

# Initial facts
facts = {"fever", "cough", "body_pain"}

# Knowledge Base: Rules
rules = [
    (["fever", "cough"], "flu"),
    (["fever", "headache"], "infection"),
    (["flu", "body_pain"], "viral_infection"),
    (["viral_infection"], "needs_rest"),
    (["infection", "high_temperature"], "serious_condition")
]

# Forward Chaining
def forward_chaining(facts):
    print("\n--- Forward Chaining ---")

    known_facts = set(facts)
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:
            if all(condition in known_facts for condition in conditions):
                if conclusion not in known_facts:
                    known_facts.add(conclusion)
                    print("Derived:", conclusion)
                    changed = True

    return known_facts


# Backward Chaining
def backward_chaining(goal, facts):
    if goal in facts:
        return True

    for conditions, conclusion in rules:
        if conclusion == goal:
            if all(backward_chaining(condition, facts)
                   for condition in conditions):
                return True

    return False


print("Initial Facts:", facts)

# Forward Chaining
result = forward_chaining(facts)
print("\nFinal Facts:", result)

# Backward Chaining
print("\n--- Backward Chaining ---")
goal = input("Enter goal : ")

if backward_chaining(goal, facts):
    print(goal, "is TRUE")
else:
    print(goal, "is FALSE")
```

```python
# An Expert System is an AI-based computer program that uses predefined knowledge and rules to solve problems or make decisions like a human expert.
# Rule-Based Expert System : It is an expert system that uses predefined IF-THEN rules to make decisions.
# A Knowledge Base is a collection of facts and rules used by an expert system to make decisions.
# An Inference Engine is the component that applies rules to known facts to derive new conclusions.
# Forward Chaining is a data-driven reasoning technique that starts with known facts and applies rules to generate new conclusions.
# Backward Chaining is a goal-driven reasoning technique that starts with a specific goal and works backward to check whether it can be proved using facts and rules.
# Recursion is a programming technique in which a function calls itself to solve a smaller part of a problem.
# set()	Stores unique facts
# list	Stores conditions in rules
# tuple	Stores conditions and conclusion together
```
